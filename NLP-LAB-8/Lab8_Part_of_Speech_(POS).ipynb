# Part-of-speech (POS)

Part-of-speech (POS) tagging is a popular NLP technique which refers to categorizing words in a text (corpus) in correspondence with a particular part of speech, depending on the definition of the word and its context.

<img src="https://drive.google.com/thumbnail?id=1-4c7pQrBjygF7bKmlrz_S_i7jlIHDYfq&sz=w1000" width="700">

## Benefits of POS tagging



- For some applications, we want to focus on certain POS:

E.g.
Nouns are important for information retrival.
Adjectives are important for sentiment analysis.

- POS Taggning is mostly useful in Named Entity Recognition (NER) applications and extracting relations between words.

- Grammar correction
- Spelling correction
- Question-Answering

### pos_tag method

````python
import nltk
# Required NLTK data for tokenization, sentence splitting, and POS tagging.
for resource in ('punkt', 'punkt_tab', 'averaged_perceptron_tagger_eng'):
    nltk.download(resource, quiet=True)
````

````python
from nltk import pos_tag
from nltk.tokenize import word_tokenize

text = "I received my new phone broken from Amazon."
tokens = word_tokenize(text)

for word in tokens:
    print(f'word: ({word}), part of speech: ({pos_tag([word])[0][1]})')
````

````python
doc = nlp(u"The quick brown fox jumped over the lazy dog's back.")
````

* To view the coarse POS tag use `token.pos_`
* To view the fine-grained tag use `token.tag_`
* To view the description of either type of tag use `spacy.explain(tag)`

````python
for token in doc:
    print(f'{token.text:{10}} {token.pos_:{8}} {token.tag_:{6}} {spacy.explain(token.tag_)}')
````

## Example: Find the adjective words in a sentance


````python
from nltk import pos_tag
from nltk.tokenize import word_tokenize, sent_tokenize
````

````python
text =  "Mobile is very handy. Packed with good performances and nice camera. This is my first iPhone. Very good experience with Apple. Battery is worst. I am hardly getting 5 hours screen on time. Positive thing is I came to know how many hours I was spending on my mobile. Because of poor battery now I am only using mobile for 4 hours a day."

print(text, end='\n\n')

for sent in sent_tokenize(text):
    tokens = word_tokenize(sent)
    part_of_speech = pos_tag(tokens)

    for i in part_of_speech:
        if 'JJ' in i:
            print(i)
    print('============================')
````

````python
# Task 1: POS information for "play" and every token
sen = nlp("I like watching football matches. I do not know how to play it though")

play_token = next(token for token in sen if token.text.lower() == "play")

print('1. Coarse-grained POS for "play":', play_token.pos_)
print('2. Fine-grained POS for "play":', play_token.tag_)
print('3. Meaning of fine-grained tag:', spacy.explain(play_token.tag_))

print('\n4. All tokens and their POS information:')
print(f'{"WORD":12} {"POS":10} {"TAG":8} EXPLANATION')
for word in sen:
    print(f'{word.text:12} {word.pos_:10} {word.tag_:8} {spacy.explain(word.tag_)}')
````

### Task 2

Compare the explanation of the words “google”  and “table” in the following sentences.
1. “Did you search on Google?”
2. “Let us google it.”
3. “The has placed the food on the table”
4. “She plans to table a motion for debate in Parliament”

**Hint:**
`word = sen[#]`


````python
# Task 2: Compare POS tagging in context
sentences = [
    "Did you search on Google?",
    "Let us google it.",
    "The has placed the food on the table",  # Original sentence as written in the lab
    "She plans to table a motion for debate in Parliament"
]

for i, sentence in enumerate(sentences, start=1):
    sen = nlp(sentence)
    print(f'\nSentence {i}: {sentence}')
    for word in sen:
        if word.text.lower() in ("google", "table"):
            print('Word:', word.text)
            print('Coarse POS:', word.pos_, '-', spacy.explain(word.pos_))
            print('Fine POS:', word.tag_, '-', spacy.explain(word.tag_))

print('\nComparison:')
print('Google is a proper noun (PROPN) when naming the search engine,')
print('but google is a verb (VERB) when it means to search online.')
print('table is a noun (NOUN) when referring to furniture,')
print('but a verb (VERB) when referring to presenting a motion.')
print('Note: Sentence 3 contains a grammatical error in the original task.')
````
